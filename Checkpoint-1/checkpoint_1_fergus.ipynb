```python
# Import the data and CSV
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("Traffic_Stops.csv")
print(f"{df.shape}, {df.head(0)}")
```

```python
# Next we look for the duplicates

# Locate time/dates that are the same, incident number?

dupes = df[df["incident_number"].duplicated()]

print(f"The number of duplicates in the dataset: {dupes.shape}") 

df_clean = df.copy()
df_clean = df_clean.drop_duplicates(subset="incident_number", ignore_index=True,keep="first")
```

```python
print(f'Cleaned data shape: {df_clean.shape}. Raw data shape: {df.shape}')
```
